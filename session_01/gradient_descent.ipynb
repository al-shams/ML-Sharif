```python
import matplotlib.pyplot as plt
import numpy as np

# داده‌ها
x = np.array([5, 3, -1, 2, 6])
y = np.array([14, 6, -5.5, 3.5, 18])

# اضافه کردن ستون 1 برای intercept
X = np.vstack([np.ones(len(x)), x]).T

# پارامترها
learning_rate = 0.01
n_iterations = 1000
n_samples = len(x)
weights = np.zeros(2)  # [w0, w1]

# Batch Gradient Descent
for _ in range(n_iterations):
    # پیش‌بینی
    y_pred = np.dot(X, weights)

    # محاسبه گرادیان
    error = y_pred - y
    gradient = (1 / n_samples) * np.dot(X.T, error)

    # به‌روزرسانی وزن‌ها
    weights = weights - learning_rate * gradient

# استخراج w0 و w1
w0 = weights[0]
w1 = weights[1]

# پیش‌بینی برای رسم نمودار
y_pred_final = np.dot(X, weights)

# رسم نمودار
plt.scatter(x, y, color="blue", label="Original data")
plt.plot(x, y_pred_final, color="red", label="Regression line")
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"Linear Regression\nw0={w0:.2f}, w1={w1:.2f}")
plt.legend()
plt.grid(True)
plt.show()

# نمایش نتایج
print(f"w0 (intercept): {w0:.4f}")
print(f"w1 (coefficient): {w1:.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# داده‌ها
x = np.array([5, 3, -1, 2, 6])
y = np.array([14, 6, -5.5, 3.5, 18])

# اضافه کردن ستون 1 برای intercept
X = np.vstack([np.ones(len(x)), x]).T

# پارامترها
learning_rate = 0.01
n_iterations = 100  # تعداد تکرارها برای هر دسته
n_samples = len(x)
batch_size = 2
weights = np.zeros(2)  # [w0, w1]

# Mini-Batch Gradient Descent
for _ in range(n_iterations):
    # تصادفی‌سازی ترتیب داده‌ها
    indices = np.random.permutation(n_samples)
    X_shuffled = X[indices]
    y_shuffled = y[indices]
    # پردازش به صورت دسته‌ای
    for i in range(0, n_samples, batch_size):
        batch_X = X_shuffled[i : i + batch_size]
        batch_y = y_shuffled[i : i + batch_size]

        # پیش‌بینی
        y_pred = np.dot(batch_X, weights)

        # محاسبه گرادیان
        error = y_pred - batch_y
        gradient = (1 / batch_size) * np.dot(batch_X.T, error)

        # به‌روزرسانی وزن‌ها
        weights = weights - learning_rate * gradient

# استخراج w0 و w1
w0 = weights[0]
w1 = weights[1]

# پیش‌بینی برای رسم نمودار
y_pred_final = np.dot(X, weights)

# رسم نمودار
plt.scatter(x, y, color="blue", label="Original data")
plt.plot(x, y_pred_final, color="red", label="Regression line")
plt.xlabel("x")
plt.ylabel("y")
plt.title(f"Mini-Batch GD\nw0={w0:.2f}, w1={w1:.2f}")
plt.legend()
plt.grid(True)
plt.show()

# نمایش نتایج
print(f"w0 (intercept): {w0:.4f}")
print(f"w1 (coefficient): {w1:.4f}")
```

```python
def gradient_descent(grad, w_init, learning_rate=0.1, n_steps=20):
    w = np.array(w_init)
    path = [w.copy()]

    for _ in range(n_steps):
        grad_w = grad(w)
        print(f"Gradient: {grad_w}")
        w = w - learning_rate * grad_w
        print(f"w: {w}")
        path.append(w.copy())

    return np.array(path)
```

```python
import matplotlib.pyplot as plt
import numpy as np


# Define the function to optimize (e.g., f(x) = x^2)
def function(x):
    return x**2


# 1. Batch Gradient Descent (original code)
def batch_gradient_descent(starting_point, learning_rate, num_iterations):
    x = starting_point
    history = [x]

    for _ in range(num_iterations):
        grad = derivative(x)
        x = x - learning_rate * grad
        history.append(x)

    return x, history


# 2. Stochastic Gradient Descent (SGD)
# For SGD, we add random noise to the gradient to simulate
# using only one sample at a time
def stochastic_gradient_descent(
    starting_point, learning_rate, num_iterations, noise_std=0.1
):
    x = starting_point
    history = [x]

    for _ in range(num_iterations):
        # Original gradient + random noise (simulate SGD)
        grad = derivative(x) + np.random.normal(0, noise_std)
        x = x - learning_rate * grad
        history.append(x)

    return x, history


# 3. Mini-batch Gradient Descent
# Simulate by averaging multiple noisy gradients
def mini_batch_gradient_descent(
    starting_point, learning_rate, num_iterations, batch_size=5, noise_std=0.1
):
    x = starting_point
    history = [x]

    for _ in range(num_iterations):
        # Compute average gradient for mini-batch
        batch_grads = []
        for _ in range(batch_size):
            grad = derivative(x) + np.random.normal(0, noise_std)
            batch_grads.append(grad)

        avg_grad = np.mean(batch_grads)
        x = x - learning_rate * avg_grad
        history.append(x)

    return x, history


# Parameters
starting_point = 10.0
learning_rate = 0.1  # reduced for better comparison
num_iterations = 50

# Set seed for reproducible results
np.random.seed(42)

# Run all three methods
print("🔴 Batch Gradient Descent:")
batch_x, batch_history = batch_gradient_descent(
    starting_point, learning_rate, num_iterations
)
print(f"Final result: x = {batch_x:.6f}, f(x) = {function(batch_x):.6f}")

print("\n🟡 Stochastic Gradient Descent:")
sgd_x, sgd_history = stochastic_gradient_descent(
    starting_point, learning_rate, num_iterations
)
print(f"Final result: x = {sgd_x:.6f}, f(x) = {function(sgd_x):.6f}")

print("\n🟢 Mini-batch Gradient Descent:")
mini_x, mini_history = mini_batch_gradient_descent(
    starting_point, learning_rate, num_iterations, batch_size=5
)
print(f"Final result: x = {mini_x:.6f}, f(x) = {function(mini_x):.6f}")

# Plotting
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Plot 1: Function and paths
x_values = np.linspace(-12, 12, 400)
y_values = function(x_values)

ax1.plot(x_values, y_values, "b-", label="f(x) = x²", linewidth=2)
ax1.plot(
    batch_history,
    [function(x) for x in batch_history],
    "r-o",
    label="Batch GD",
    markersize=4,
    alpha=0.7,
)
ax1.plot(
    sgd_history,
    [function(x) for x in sgd_history],
    "y-s",
    label="SGD",
    markersize=3,
    alpha=0.7,
)
ax1.plot(
    mini_history,
    [function(x) for x in mini_history],
    "g-^",
    label="Mini-batch GD",
    markersize=3,
    alpha=0.7,
)

ax1.set_xlabel("x")
ax1.set_ylabel("f(x)")
ax1.set_title("Convergence Paths Comparison")
ax1.legend()
ax1.grid(True, alpha=0.3)

# Plot 2: Convergence over iterations
iterations = range(len(batch_history))
ax2.plot(
    iterations, [abs(x) for x in batch_history], "r-", label="Batch GD", linewidth=2
)
ax2.plot(iterations, [abs(x) for x in sgd_history], "y-", label="SGD", linewidth=2)
ax2.plot(
    iterations, [abs(x) for x in mini_history], "g-", label="Mini-batch GD", linewidth=2
)

ax2.set_xlabel("Iterations")
ax2.set_ylabel("|x| (distance from optimum)")
ax2.set_title("Convergence Speed")
ax2.set_yscale("log")  # logarithmic scale for better visualization
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Final Comparison
print("\n" + "=" * 50)
print("📊 Final Comparison:")
print(f"Batch GD:      |x| = {abs(batch_x):.6f}")
print(f"SGD:           |x| = {abs(sgd_x):.6f}")
print(f"Mini-batch GD: |x| = {abs(mini_x):.6f}")
print("=" * 50)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate synthetic data for linear regression: y = 2x + 3 + noise
np.random.seed(42)
X = np.random.rand(100, 1) * 10  # 100 points between 0 and 10
y = 2 * X + 3 + np.random.randn(100, 1) * 0.5  # y = 2x + 3 + noise


# Define the cost function (Mean Squared Error)
def compute_cost(X, y, theta):
    m = len(y)
    predictions = X @ theta
    cost = (1 / (2 * m)) * np.sum((predictions - y) ** 2)
    return cost


# Batch Gradient Descent
def batch_gradient_descent(X, y, theta_init, learning_rate, num_iterations):
    m = len(y)
    theta = theta_init.copy()
    history = [compute_cost(X, y, theta)]

    for _ in range(num_iterations):
        predictions = X @ theta
        gradient = (1 / m) * X.T @ (predictions - y)
        theta -= learning_rate * gradient
        history.append(compute_cost(X, y, theta))

    return theta, history


# Stochastic Gradient Descent
def stochastic_gradient_descent(X, y, theta_init, learning_rate, num_iterations):
    m = len(y)
    theta = theta_init.copy()
    history = [compute_cost(X, y, theta)]

    for i in range(num_iterations):
        for i in range(m):
            xi = X[i : i + 1]
            yi = y[i : i + 1]
            prediction = xi @ theta
            gradient = xi.T @ (prediction - yi)
            theta -= learning_rate * gradient
        history.append(compute_cost(X, y, theta))

    return theta, history


# Mini-Batch Gradient Descent
def mini_batch_gradient_descent(
    X, y, theta_init, learning_rate, num_iterations, batch_size
):
    m = len(y)
    theta = theta_init.copy()
    history = [compute_cost(X, y, theta)]

    for _ in range(num_iterations):
        indices = np.random.permutation(m)
        X_shuffled = X[indices]
        y_shuffled = y[indices]

        for i in range(0, m, batch_size):
            xi = X_shuffled[i : i + batch_size]
            yi = y_shuffled[i : i + batch_size]
            prediction = xi @ theta
            gradient = (1 / batch_size) * xi.T @ (prediction - yi)
            theta -= learning_rate * gradient
        history.append(compute_cost(X, y, theta))

    return theta, history


# Prepare data with bias term (add column of ones for intercept)
X_b = np.c_[np.ones((100, 1)), X]  # Add bias term
theta_init = np.random.randn(2, 1)  # Initial parameters [theta0, theta1]

# Parameters
learning_rate = 0.01
num_iterations = 100
batch_size = 10  # For Mini-Batch GD

# Run all three methods
theta_batch, cost_history_batch = batch_gradient_descent(
    X_b, y, theta_init, learning_rate, num_iterations
)
theta_sgd, cost_history_sgd = stochastic_gradient_descent(
    X_b, y, theta_init, learning_rate, num_iterations
)
theta_mini, cost_history_mini = mini_batch_gradient_descent(
    X_b, y, theta_init, learning_rate, num_iterations, batch_size
)

# Plot cost history for comparison
plt.figure(figsize=(12, 6))
plt.plot(cost_history_batch, label="Batch Gradient Descent", linewidth=2)
plt.plot(cost_history_sgd, label="Stochastic Gradient Descent", linewidth=2)
plt.plot(
    cost_history_mini, label="Mini-Batch Gradient Descent (batch_size=10)", linewidth=2
)
plt.xlabel("Iteration")
plt.ylabel("Cost (MSE)")
plt.title("Cost Function Convergence for Different Gradient Descent Methods")
plt.legend()
plt.grid(True)
plt.show()

# Print final parameters
print("Batch GD - Final theta:", theta_batch.ravel())
print("SGD - Final theta:", theta_sgd.ravel())
print("Mini-Batch GD - Final theta:", theta_mini.ravel())

# Plot the regression lines
plt.figure(figsize=(12, 6))
plt.scatter(X, y, color="blue", label="Data points")
x_range = np.array([[1, 0], [1, 10]])
plt.plot(x_range[:, 1], x_range @ theta_batch, "r-", label="Batch GD")
plt.plot(x_range[:, 1], x_range @ theta_sgd, "g--", label="SGD")
plt.plot(x_range[:, 1], x_range @ theta_mini, "m-.", label="Mini-Batch GD")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression Fit with Different Gradient Descent Methods")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np


# Define the function to optimize (e.g., f(x) = x^2)
def function(x):
    return x**2


# Batch Gradient Descent
def batch_gradient_descent(starting_point, learning_rate, num_iterations):
    x = starting_point
    history_x = [x]
    history_cost = [function(x)]

    for _ in range(num_iterations):
        grad = derivative(x)
        x = x - learning_rate * grad
        history_x.append(x)
        history_cost.append(function(x))

    return x, history_x, history_cost


# Stochastic Gradient Descent (simulated with noise)
def stochastic_gradient_descent(starting_point, learning_rate, num_iterations):
    x = starting_point
    history_x = [x]
    history_cost = [function(x)]

    for _ in range(num_iterations):
        # Simulate stochasticity by adding small noise to gradient
        grad = derivative(x) + np.random.normal(0, 0.1)
        x = x - learning_rate * grad
        history_x.append(x)
        history_cost.append(function(x))

    return x, history_x, history_cost


# Mini-Batch Gradient Descent (simulated with batches)
def mini_batch_gradient_descent(
    starting_point, learning_rate, num_iterations, batch_size=5
):
    x = starting_point
    history_x = [x]
    history_cost = [function(x)]

    for _ in range(num_iterations):
        # Simulate mini-batch by averaging gradients with noise
        for _ in range(batch_size):
            grad = derivative(x) + np.random.normal(0, 0.05)
            x = x - (learning_rate / batch_size) * grad
        history_x.append(x)
        history_cost.append(function(x))

    return x, history_x, history_cost


# Parameters
starting_point = 10.0
learning_rate = 0.1
num_iterations = 50
batch_size = 5

# Run all three methods
x_batch, history_x_batch, history_cost_batch = batch_gradient_descent(
    starting_point, learning_rate, num_iterations
)
x_sgd, history_x_sgd, history_cost_sgd = stochastic_gradient_descent(
    starting_point, learning_rate, num_iterations
)
x_mini, history_x_mini, history_cost_mini = mini_batch_gradient_descent(
    starting_point, learning_rate, num_iterations, batch_size
)

# Generate points for plotting the function
x_values = np.linspace(-10, 10, 400)
y_values = function(x_values)

# Plot separate figures for each method
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))

# Batch Gradient Descent Plot
ax1.plot(x_values, y_values, "b-", label="f(x) = x^2")
ax1.plot(
    history_x_batch,
    [function(x) for x in history_x_batch],
    "ro-",
    label="Batch GD Path",
)
ax1.set_xlabel("x")
ax1.set_ylabel("f(x)")
ax1.set_title("Batch Gradient Descent")
ax1.legend()
ax1.grid(True)

# Stochastic Gradient Descent Plot
ax2.plot(x_values, y_values, "b-", label="f(x) = x^2")
ax2.plot(history_x_sgd, [function(x) for x in history_x_sgd], "go-", label="SGD Path")
ax2.set_xlabel("x")
ax2.set_ylabel("f(x)")
ax2.set_title("Stochastic Gradient Descent")
ax2.legend()
ax2.grid(True)

# Mini-Batch Gradient Descent Plot
ax3.plot(x_values, y_values, "b-", label="f(x) = x^2")
ax3.plot(
    history_x_mini,
    [function(x) for x in history_x_mini],
    "mo-",
    label="Mini-Batch GD Path",
)
ax3.set_xlabel("x")
ax3.set_ylabel("f(x)")
ax3.set_title("Mini-Batch Gradient Descent")
ax3.legend()
ax3.grid(True)

plt.tight_layout()
plt.show()

# Plot cost history for each method
plt.figure(figsize=(10, 6))
plt.plot(history_cost_batch, "r-", label="Batch Gradient Descent")
plt.plot(history_cost_sgd, "g--", label="Stochastic Gradient Descent")
plt.plot(history_cost_mini, "m-.", label="Mini-Batch Gradient Descent")
plt.xlabel("Iteration")
plt.ylabel("Cost (f(x))")
plt.title("Cost Function Convergence for Gradient Descent Methods")
plt.legend()
plt.grid(True)
plt.show()

# Print final results
print(f"Batch GD - Optimal x: {x_batch:.4f}, f(x): {function(x_batch):.4f}")
print(f"SGD - Optimal x: {x_sgd:.4f}, f(x): {function(x_sgd):.4f}")
print(f"Mini-Batch GD - Optimal x: {x_mini:.4f}, f(x): {function(x_mini):.4f}")
```
